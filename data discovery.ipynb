```python
import pandas as pd

df = pd.read_csv('raw_job_postings.csv')

# 1. Unique job_id, created once, shared across all bridge tables
df.insert(0, 'job_id', range(1, len(df) + 1))

# 2. Fact table - one row per posting
df.to_csv('job_postings_fact.csv', index=False)

# --- Bridge table: skills ---
skills_df = df[['job_id', 'Required Skills']].copy()
skills_df['Required Skills'] = skills_df['Required Skills'].str.split(',')
skills_df = skills_df.explode('Required Skills')
skills_df['Required Skills'] = skills_df['Required Skills'].str.strip()
skills_df = skills_df.rename(columns={'Required Skills': 'skill'})
skills_df.to_csv('job_skills_bridge.csv', index=False)

# --- Bridge table: programming languages ---
lang_df = df[['job_id', 'Programming Languages Required']].copy()
lang_df['Programming Languages Required'] = lang_df['Programming Languages Required'].str.split(',')
lang_df = lang_df.explode('Programming Languages Required')
lang_df['Programming Languages Required'] = lang_df['Programming Languages Required'].str.strip()
lang_df = lang_df.rename(columns={'Programming Languages Required': 'language'})
lang_df.to_csv('job_languages_bridge.csv', index=False)
```

```python
import pandas as pd
import re

# ============================================================
# 1. LOAD DATA
# ============================================================

input_file = "job_postings_fact.csv"
output_file = "job_postings_salary_cleaned.csv"

df = pd.read_csv(input_file, low_memory=False)

print("Original shape:", df.shape)
print("Columns:")
print(df.columns.tolist())


# ============================================================
# 2. MAKE SURE JOB ID IS PRESERVED
# ============================================================

# If your Job ID column has a different name, change it here.
# For example: "job_id", "Job ID", "id", etc.

if "Job ID" in df.columns:
    print("Job ID column found.")
elif "job_id" in df.columns:
    df.rename(columns={"job_id": "Job ID"}, inplace=True)
    print("job_id renamed to Job ID.")
else:
    print("WARNING: No Job ID column found.")


# ============================================================
# 3. KEEP ORIGINAL SALARY RANGE
# ============================================================

df["Salary Range Original"] = df["Salary Range"]


# ============================================================
# 4. FIX MANGLED ENCODING
# ============================================================

def fix_encoding(text):
    if pd.isna(text):
        return text

    text = str(text)

    try:
        return text.encode("latin1").decode("utf-8")
    except (UnicodeDecodeError, UnicodeEncodeError):
        return text


df["Salary Range"] = df["Salary Range"].apply(fix_encoding)


# ============================================================
# 5. REMOVE JUNK SUFFIXES
# ============================================================

# Examples:
# "$80,000 - $100,000 + 1.0% Equity"
# "$80,000-$100,000 + Equity"
# etc.

df["Salary Range"] = (
    df["Salary Range"]
    .astype("string")
    .str.replace(r"\s*\+.*$", "", regex=True)
    .str.strip()
)


# ============================================================
# 6. CURRENCY DETECTION
# ============================================================

currency_map = {
    r"^\$": "USD",
    r"^£": "GBP",
    r"^€": "EUR",
    r"^¥": "JPY",
    r"^₹": "INR",
    r"^S\$": "SGD",
    r"^A\$": "AUD",
    r"^C\$": "CAD",
    r"^CHF": "CHF",
}


def get_currency(text):

    if pd.isna(text):
        return None

    text = str(text).strip()

    for pattern, code in currency_map.items():
        if re.match(pattern, text):
            return code

    return "UNKNOWN"


df["currency"] = df["Salary Range"].apply(get_currency)


# ============================================================
# 7. PARSE INDIVIDUAL SALARY AMOUNT
# ============================================================

def parse_amount(token):

    if pd.isna(token):
        return None

    token = str(token).strip()

    # Detect lakh notation
    # Example: 10L = 1,000,000
    is_lakh = token.upper().endswith("L")

    # Remove everything except digits and decimal point
    token = re.sub(r"[^\d.]", "", token)

    if token == "":
        return None

    try:
        value = float(token)
    except ValueError:
        return None

    if is_lakh:
        value *= 100_000

    return value


# ============================================================
# 8. PARSE SALARY RANGE
# ============================================================

def parse_range(text):

    if pd.isna(text):
        return None, None

    text = str(text).strip()

    # Remove currency symbols
    text = re.sub(r"^(S\$|A\$|C\$|CHF|\$|£|€|¥|₹)\s*", "", text)

    # Remove possible second currency symbols
    text = re.sub(r"(S\$|A\$|C\$|CHF|\$|£|€|¥|₹)", "", text)

    # Split into minimum and maximum
    parts = re.split(r"\s*-\s*", text)

    if len(parts) != 2:
        return None, None

    low = parse_amount(parts[0])
    high = parse_amount(parts[1])

    return low, high


# ============================================================
# 9. CREATE MIN/MAX SALARY COLUMNS
# ============================================================

df[["salary_min", "salary_max"]] = df["Salary Range"].apply(
    lambda x: pd.Series(parse_range(x))
)


# ============================================================
# 10. CALCULATE AVERAGE SALARY
# ============================================================

df["salary_avg"] = (
    df["salary_min"] + df["salary_max"]
) / 2


# ============================================================
# 11. SALARY PARSING STATUS
# ============================================================

df["salary_parsing_status"] = "Parsed"

df.loc[
    df["salary_min"].isna() | df["salary_max"].isna(),
    "salary_parsing_status"
] = "Not Parsed"


# ============================================================
# 12. SANITY CHECK
# ============================================================

print("\n==============================")
print("CURRENCY DISTRIBUTION")
print("==============================")

print(df["currency"].value_counts(dropna=False))


print("\n==============================")
print("SALARY SAMPLE")
print("==============================")

columns_to_show = [
    col for col in [
        "Job ID",
        "Salary Range Original",
        "Salary Range",
        "currency",
        "salary_min",
        "salary_max",
        "salary_avg",
        "salary_parsing_status"
    ]
    if col in df.columns
]

print(df[columns_to_show].head(10))


print("\n==============================")
print("PARSING RESULTS")
print("==============================")

print(
    "Unparsed rows:",
    df["salary_min"].isna().sum()
)

print(
    "Parsed rows:",
    df["salary_min"].notna().sum()
)


# ============================================================
# 13. SAVE TO NEW EXCEL FILE
# ============================================================

# Reorder columns so Job ID is first

if "Job ID" in df.columns:

    other_columns = [
        col for col in df.columns
        if col != "Job ID"
    ]

    df = df[
        ["Job ID"] + other_columns
    ]


# Save as Excel
df.to_csv(
    output_file,
    index=False)


print("\n==============================")
print("DONE")
print("==============================")

print("Output file:", output_file)
print("Final shape:", df.shape)
```
